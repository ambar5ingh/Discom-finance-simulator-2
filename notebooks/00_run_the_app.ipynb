# DISCOM Finance Simulator: run the app in Jupyter

**Inside Jupyter:** run all cells (*Run → Run All Cells*). The app appears below the last cell, with sliders on the left and results on the right.

**As a web page:** from the repository folder, run `panel serve notebooks/00_run_the_app.ipynb --show`. Panel serves this notebook's output as a standalone app.

```python
import runpy
from pathlib import Path

import panel as pn

pn.extension("tabulator", sizing_mode="stretch_width")

# works whether Jupyter was started in the repository root or in notebooks/
ROOT = next(p for p in [Path.cwd(), Path.cwd().parent] if (p / "app" / "app.py").exists())
ns = runpy.run_path(str(ROOT / "app" / "app.py"), run_name="discom_notebook")
```

```python
app = pn.Row(
    pn.Column("## DISCOM Finance Simulator", *ns["sidebar"], width=360),
    ns["tabs"],
)
app.servable(title="DISCOM Finance Simulator")
app
```
